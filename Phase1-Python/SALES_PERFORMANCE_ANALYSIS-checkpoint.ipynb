```python
# Feature Engineering

'''In this section, we create business metrics required for analysis.
These metrics are not available in the raw dataset and are calculated
using business rules.'''
```

```python
# Gross Sales before discount
sales["GrossSales"] = sales["Quantity"] * sales["UnitPrice"]

# Discount Amount
sales["DiscountAmount"] = (
    sales["GrossSales"] * sales["DiscountPercent"] / 100
)

# Net Sales after discount
sales["NetSales"] = (
    sales["GrossSales"] - sales["DiscountAmount"]
)

# Total Cost
sales["TotalCost"] = (
    sales["Quantity"] * sales["UnitCost"]
)

# Profit
sales["Profit"] = (
    sales["NetSales"] - sales["TotalCost"]
)

# Profit Margin (%)
sales["ProfitMargin"] = np.where(
    sales["NetSales"] == 0,
    0,
    (sales["Profit"] / sales["NetSales"]) * 100
)
```

```python
sales["Year"] = sales["OrderDate"].dt.year

sales["Quarter"] = sales["OrderDate"].dt.quarter

sales["Month"] = sales["OrderDate"].dt.month

sales["MonthName"] = sales["OrderDate"].dt.strftime("%B")

sales["Day"] = sales["OrderDate"].dt.day

sales["Weekday"] = sales["OrderDate"].dt.day_name()
```

```python
sales["AgeGroup"] = pd.cut(
    sales["Age"],
    bins=[0, 25, 35, 45, 55, 100],
    labels=[
        "18-25",
        "26-35",
        "36-45",
        "46-55",
        "55+"
    ]
)
```

```python
# Executive KPI Dashboard

'''This section calculates the key performance indicators (KPIs)
used by management to evaluate sales performance, profitability,
customer behavior, and operational efficiency.'''
```

```python
total_revenue = sales["NetSales"].sum()

total_profit = sales["Profit"].sum()

total_orders = sales["OrderID"].nunique()

total_customers = sales["CustomerID"].nunique()

total_products = sales["ProductID"].nunique()

units_sold = sales["Quantity"].sum()

avg_order_value = sales.groupby("OrderID")["NetSales"].sum().mean()

avg_discount = sales["DiscountPercent"].mean()

profit_margin = (total_profit / total_revenue) * 100
```

```python
# DISPLAY KPIs

print("="*60)
print("EXECUTIVE KPI SUMMARY")
print("="*60)

print(f"Total Revenue       : ₹{total_revenue:,.2f}")
print(f"Total Profit        : ₹{total_profit:,.2f}")
print(f"Profit Margin       : {profit_margin:.2f}%")
print(f"Total Orders        : {total_orders:,}")
print(f"Total Customers     : {total_customers:,}")
print(f"Total Products Sold : {total_products:,}")
print(f"Units Sold          : {units_sold:,}")
print(f"Average Order Value : ₹{avg_order_value:,.2f}")
print(f"Average Discount    : {avg_discount:.2f}%")
```

```python
# MONTHLY SALES TREND

monthly_sales = (
    sales.groupby(["Year", "Month", "MonthName"], as_index=False)
    .agg(
        TotalRevenue=("NetSales", "sum"),
        TotalProfit=("Profit", "sum"),
        Orders=("OrderID", "nunique")
    )
)

monthly_sales = monthly_sales.sort_values(["Year", "Month"])

monthly_sales.head()
```

```python
# MONTHLY REVENUE

plt.figure(figsize=(14,6))

plt.plot(
    monthly_sales["MonthName"],
    monthly_sales["TotalRevenue"],
    marker="o",
    markersize=8,
    linewidth=3
)

plt.title(
    "Monthly Revenue Trend",
    fontsize=18,
    fontweight="bold",
    pad=20
)

plt.xlabel(
    "Month",
    fontsize=13,
    fontweight="bold"
)

plt.ylabel(
    "Revenue (₹)",
    fontsize=13,
    fontweight="bold"
)

plt.xticks(rotation=45, fontsize=11)

plt.yticks(fontsize=11)

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()
```

```python
# MONTHLY PROFIT TREND

plt.figure(figsize=(14,6))

plt.plot(
    monthly_sales["MonthName"],
    monthly_sales["TotalProfit"],
    color="green",
    marker="o",
    markersize=8,
    linewidth=3,
    label="Profit"
)

# Highlight Highest Profit
max_idx = monthly_sales["TotalProfit"].idxmax()
plt.scatter(
    monthly_sales.loc[max_idx, "MonthName"],
    monthly_sales.loc[max_idx, "TotalProfit"],
    color="darkgreen",
    s=180,
    zorder=5,
    label="Highest Profit"
)

# Highlight Lowest Profit
min_idx = monthly_sales["TotalProfit"].idxmin()
plt.scatter(
    monthly_sales.loc[min_idx, "MonthName"],
    monthly_sales.loc[min_idx, "TotalProfit"],
    color="red",
    s=180,
    zorder=5,
    label="Lowest Profit"
)

# Data Labels
for x, y in zip(monthly_sales["MonthName"], monthly_sales["TotalProfit"]):
    plt.text(
        x,
        y,
        f"{y/1_000_000:.1f}M",
        ha="center",
        va="bottom",
        fontsize=9
    )

plt.title(
    "Monthly Profit Trend",
    fontsize=18,
    fontweight="bold",
    pad=20
)

plt.xlabel(
    "Month",
    fontsize=13,
    fontweight="bold"
)

plt.ylabel(
    "Profit (₹)",
    fontsize=13,
    fontweight="bold"
)

plt.xticks(rotation=45, fontsize=11)
plt.yticks(fontsize=11)

plt.grid(alpha=0.3)

plt.legend()

plt.tight_layout()

plt.show()
```

```python
# DISCOUNT VS PROFIT ANALYSIS

sales["DiscountCategory"] = pd.cut(
    sales["DiscountPercent"],
    bins=[-1, 0, 5, 10, 15, 20, 100],
    labels=[
        "0%",
        "1-5%",
        "6-10%",
        "11-15%",
        "16-20%",
        "20%+"
    ]
)
```

```python
# ANALYZE REVENUE AND PROFIT

discount_analysis = (
    sales.groupby("DiscountCategory", observed=False)
    .agg(
        Revenue=("NetSales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("OrderID", "nunique")
    )
    .reset_index()
)

discount_analysis
```
